```
# Nguyen Tat Tan Tai - 24521555

"""
Temporal Convolutional Network (TCN) for hospital admissions prediction.
This script builds a TCN to forecast admissions using air pollution data.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, LabelBinarizer
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, Conv1D, Add, Activation, GlobalAveragePooling1D
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

# ===========================================================================
# 1. Configuration
# ===========================================================================
DATA_FILE = "merged_project_data.csv"       # Your merged CSV file
LOOKBACK = 7                                # Number of past days to look back
TEST_SPLIT = 0.2                            # 20% test data
EPOCHS = 100
BATCH_SIZE = 32
PATIENCE = 15
LEARNING_RATE = 0.001
DILATIONS = [1, 2, 4, 8, 16]               # Dilation rates
FILTERS = 64
KERNEL_SIZE = 3
DROPOUT_RATE = 0.2

FEATURE_COLS = ['pm25', 'pm10', 'o3', 'no2', 'so2', 'co']
TARGET_COL = 'hospital_admissions'
CITY_COL = 'city'

# ===========================================================================
# 2. Load and clean data
# ===========================================================================
df = pd.read_csv(DATA_FILE)
print(f"Raw data shape: {df.shape}")

# Keep only needed columns and drop missing values
df_clean = df[FEATURE_COLS + [TARGET_COL, CITY_COL]].dropna()
print(f"After cleaning: {len(df_clean)} rows")

# ===========================================================================
# 3. One-hot encode city and combine features
# ===========================================================================
encoder = LabelBinarizer()
city_ohe = encoder.fit_transform(df_clean[CITY_COL])
print(f"Cities: {encoder.classes_}")

X_raw = np.hstack([df_clean[FEATURE_COLS].values, city_ohe])   # (n, 9)
y_raw = df_clean[TARGET_COL].values.reshape(-1, 1)

# Normalize data to [0, 1]
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()
X_scaled = scaler_X.fit_transform(X_raw)
y_scaled = scaler_y.fit_transform(y_raw)

# ===========================================================================
# 4. Create sequences per city (avoid crossing city boundaries)
# ===========================================================================
def create_sequences_per_city(df, city_col, X_scaled, y_scaled, lookback):
    X_seq, y_seq = [], []
    for city in df[city_col].unique():
        # Get indices for this city
        idx = df[df[city_col] == city].index
        temp = pd.DataFrame(X_scaled, columns=[f'f{i}' for i in range(X_scaled.shape[1])])
        temp[city_col] = df[city_col].values
        temp['y'] = y_scaled.flatten()
        city_temp = temp[temp[city_col] == city].reset_index(drop=True)
        X_city = city_temp.iloc[:, :-2].values
        y_city = city_temp['y'].values.reshape(-1, 1)
        for i in range(len(X_city) - lookback):
            X_seq.append(X_city[i:i+lookback])
            y_seq.append(y_city[i+lookback])
    return np.array(X_seq), np.array(y_seq)

X_seq, y_seq = create_sequences_per_city(df_clean, CITY_COL, X_scaled, y_scaled, LOOKBACK)
print(f"Total sequences: {X_seq.shape[0]}, shape: {X_seq.shape}")

# ===========================================================================
# 5. Train/test split (chronological)
# ===========================================================================
split = int((1 - TEST_SPLIT) * len(X_seq))
X_train, X_test = X_seq[:split], X_seq[split:]
y_train, y_test = y_seq[:split], y_seq[split:]
print(f"Train: {X_train.shape[0]}, Test: {X_test.shape[0]}")

# ===========================================================================
# 6. Build TCN model
# ===========================================================================
def residual_block(x, filters, kernel_size, dilation_rate, dropout_rate):
    conv = Conv1D(filters=filters,
                  kernel_size=kernel_size,
                  padding='causal',         # no future information
                  dilation_rate=dilation_rate,
                  activation='relu')(x)
    conv = Dropout(dropout_rate)(conv)
    if x.shape[-1] != filters:
        x = Conv1D(filters=filters, kernel_size=1, padding='same')(x)
    out = Add()([x, conv])
    out = Activation('relu')(out)
    return out

input_dim = X_seq.shape[2]
inputs = Input(shape=(LOOKBACK, input_dim))
x = inputs
for d in DILATIONS:
    x = residual_block(x, FILTERS, KERNEL_SIZE, d, DROPOUT_RATE)
x = GlobalAveragePooling1D()(x)
outputs = Dense(1)(x)

model = Model(inputs, outputs)
model.compile(optimizer=Adam(learning_rate=LEARNING_RATE), loss='mse', metrics=['mae'])
model.summary()

# ===========================================================================
# 7. Train
# ===========================================================================
early_stop = EarlyStopping(monitor='val_loss', patience=PATIENCE, restore_best_weights=True)
history = model.fit(X_train, y_train,
                    validation_split=0.2,
                    epochs=EPOCHS,
                    batch_size=BATCH_SIZE,
                    callbacks=[early_stop],
                    verbose=1)

# ===========================================================================
# 8. Evaluate
# ===========================================================================
y_pred_scaled = model.predict(X_test, verbose=0)
y_test_orig = scaler_y.inverse_transform(y_test)
y_pred_orig = scaler_y.inverse_transform(y_pred_scaled)

mae = mean_absolute_error(y_test_orig, y_pred_orig)
rmse = np.sqrt(mean_squared_error(y_test_orig, y_pred_orig))
r2 = r2_score(y_test_orig, y_pred_orig)

print("\n=== TCN Results ===")
print(f"MAE:  {mae:.4f} patients/day")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")

# MAPE (skip zeros)
mask = (y_test_orig.flatten() != 0)
if mask.any():
    mape = np.mean(np.abs((y_test_orig[mask] - y_pred_orig[mask]) / y_test_orig[mask])) * 100
    print(f"MAPE: {mape:.2f}%")

# ===========================================================================
# 9. Plots (optional)
# ===========================================================================
plt.figure(figsize=(10, 4))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('TCN Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE')
plt.legend()
plt.show()

plt.figure(figsize=(12, 5))
plt.plot(y_test_orig[:200], label='Actual', alpha=0.7)
plt.plot(y_pred_orig[:200], label='TCN Predicted', alpha=0.7)
plt.title('TCN: Actual vs Predicted (Test Set)')
plt.xlabel('Test sample')
plt.ylabel('Admissions')
plt.legend()
plt.show()
```
